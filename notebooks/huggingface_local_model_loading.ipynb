# Loading a Local Hugging Face Model

This notebook is a walkthrough for configuring a basic Hugging Face text-generation model in `local-ml-vibes`. It is intentionally safe to open and run: the cell that would download or initialize a model is commented out.

The notebook demonstrates configuration only. Uncomment the loading cell when you are ready to install the optional model dependencies and explicitly permit model initialization.

## 1. Install optional dependencies

From the repository root, run this in a terminal before enabling model loading:

```bash
uv sync --extra models
```

This notebook does not install packages automatically.

````python
from pathlib import Path

from local_ml_vibes.models.config import ModelConfig

MODEL_ID = "distilbert/distilgpt2"
CACHE_DIR = Path("models/huggingface")

config = ModelConfig(
    repository_id=MODEL_ID,
    cache_dir=CACHE_DIR,
    local_files_only=False,
    device="auto",
    trust_remote_code=False,
)

config.validate()
config
````

## 2. Choose a model deliberately

`distilbert/distilgpt2` is a small text-generation example, not a recommendation for production use. Pin a `revision` when reproducibility matters. Use `local_files_only=True` only after the required files already exist in the cache.

Other example model IDs to consider, depending on the task and available hardware:

```python
# MODEL_ID = "gpt2"
# MODEL_ID = "HuggingFaceTB/SmolLM2-135M-Instruct"
# MODEL_ID = "Qwen/Qwen2.5-0.5B-Instruct"
```

Review each model's license, memory requirements, intended use, and security considerations before loading it. Keep `trust_remote_code=False` unless the model specifically requires custom code and you have reviewed that code.

````python
# Safe inspection: this prints configuration only and does not contact Hugging Face.
print(config)
print(f"Cache directory: {config.cache_dir}")
print(f"Local files only: {config.local_files_only}")
````

## 4. Local-only follow-up

After a model has been downloaded and reviewed, a separate local-only configuration can prevent network access:

```python
local_config = ModelConfig(
    repository_id=MODEL_ID,
    revision="replace-with-a-reviewed-commit-or-tag",
    cache_dir=CACHE_DIR,
    local_files_only=True,
    trust_remote_code=False,
)
```

The repository's loader validates configuration before importing Transformers. It does not load anything merely because the package or notebook is imported.
